```python
import os
from dotenv import load_dotenv

from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_core.prompts import ChatPromptTemplate
```

```python
load_dotenv()

api_key = os.getenv("OPENROUTER_API_KEY")

if api_key:
    print("OpenRouter API Key Found")
else:
    print("OpenRouter API Key Not Found")
```

```python
llm = ChatOpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=api_key,
    model="openai/gpt-oss-20b",
    temperature=0
)
```

```python
response = llm.invoke([
    HumanMessage(
        content="What is an Enterprise AI Assistant? Explain in two sentences."
    )
])

print(response.content)
```

### ChatPromptTemplate

A chat prompt template creates a reusable prompt structure with placeholders for dynamic information.

Here, the employee's question is represented using `{question}`.

```python
prompt = ChatPromptTemplate.from_template(
    """You are an AI assistant for NovaTech Solutions Pvt. Ltd.

Answer the employee's question clearly and professionally.

Employee question:
{question}
"""
)
```

```python
prompt_message = prompt.invoke({
    "question": "What are the responsibilities of the HR department?"
})

response = llm.invoke(prompt_message)

print(response.content)
```

### Simple LCEL Chain

LangChain Expression Language allows the prompt and language model to be connected using the pipe operator.

The flow is:

Prompt → LLM → Response

```python
chain = prompt | llm
```

```python
response = chain.invoke({
    "question": "What are the responsibilities of the HR department?"
})

print(response.content)
```
