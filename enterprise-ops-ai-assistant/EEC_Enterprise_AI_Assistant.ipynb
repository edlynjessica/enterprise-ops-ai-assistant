```python
from pathlib import Path
from reportlab.lib.pagesizes import A4
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.lib.styles import getSampleStyleSheet

documents_dir = Path("documents")
documents_dir.mkdir(exist_ok=True)

styles = getSampleStyleSheet()


def create_pdf(filename, title, content):
    file_path = documents_dir / filename

    doc = SimpleDocTemplate(
        str(file_path),
        pagesize=A4
    )

    story = [
        Paragraph(title, styles["Title"]),
        Spacer(1, 20)
    ]

    for paragraph in content:
        story.append(
            Paragraph(paragraph, styles["BodyText"])
        )
        story.append(Spacer(1, 10))

    doc.build(story)

    print(f"Created: {file_path}")
```

```python
create_pdf(
    "leave_policy.pdf",
    "NovaTech Solutions Pvt. Ltd. - Employee Leave Policy",
    [
        "Employees are entitled to 24 days of paid leave per calendar year.",
        "Employees should submit planned leave requests at least 3 working days in advance.",
        "All planned leave requests must be approved by the employee's reporting manager.",
        "Employees may use sick leave when they are unable to work due to illness.",
        "Casual leave may be used for personal or short-notice requirements.",
        "For emergency or sick leave, employees should inform their manager as soon as possible.",
        "Unused annual leave may be carried forward up to 10 days into the next calendar year.",
        "The HR department maintains official employee leave records."
    ]
)

create_pdf(
    "attendance_policy.pdf",
    "NovaTech Solutions Pvt. Ltd. - Attendance Policy",
    [
        "Standard working hours are 9:00 AM to 6:00 PM, Monday to Friday.",
        "Employees are expected to record attendance through the company's attendance system.",
        "Employees arriving more than 15 minutes late should inform their reporting manager.",
        "Missing attendance records must be regularized within 2 working days.",
        "Employees should maintain accurate attendance records.",
        "Repeated attendance violations may be reviewed by the HR department."
    ]
)

create_pdf(
    "wfh_policy.pdf",
    "NovaTech Solutions Pvt. Ltd. - Work From Home Policy",
    [
        "Eligible employees may work from home up to 2 days per week.",
        "Work from home requires prior approval from the reporting manager.",
        "Employees working remotely must remain available during standard working hours.",
        "Employees must attend scheduled meetings and remain reachable through official communication channels.",
        "Employees must maintain a suitable working environment and reliable internet connection.",
        "Employees are responsible for completing their assigned work while working remotely."
    ]
)

create_pdf(
    "employee_handbook.pdf",
    "NovaTech Solutions Pvt. Ltd. - Employee Handbook",
    [
        "NovaTech Solutions promotes a professional, respectful, and collaborative workplace.",
        "Employees are expected to follow company policies and maintain professional conduct.",
        "Company systems and data must be used only for authorized business purposes.",
        "Employees must protect confidential company information.",
        "The HR department can be contacted for employee policy questions.",
        "Standard office working hours are 9:00 AM to 6:00 PM, Monday to Friday.",
        "Employees should communicate respectfully with colleagues, managers, and business partners."
    ]
)
```

### PDF Loader

`PyPDFLoader` loads PDF documents into LangChain `Document` objects. Each page becomes a document containing the page text and metadata such as the source filename and page number.

```python
from langchain_community.document_loaders import PyPDFLoader

pdf_files = list(documents_dir.glob("*.pdf"))

documents = []

for pdf_file in pdf_files:
    loader = PyPDFLoader(str(pdf_file))
    documents.extend(loader.load())

print("PDF files:", len(pdf_files))
print("Total pages:", len(documents))

for doc in documents:
    print(
        doc.metadata.get("source"),
        "-> page",
        doc.metadata.get("page")
    )
```

### Text Splitting

Text splitting divides large documents into smaller chunks so that relevant sections can be retrieved efficiently.

We use `RecursiveCharacterTextSplitter` with a chunk size of 800 characters and an overlap of 100 characters.

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=100
)

chunks = text_splitter.split_documents(documents)

print("Original pages:", len(documents))
print("Generated chunks:", len(chunks))
```

### Embeddings

Embeddings convert text into numerical vectors that represent semantic meaning.

The embedding flow is:

Document Text → Embedding Model → Numerical Vector

We use the lightweight local `all-MiniLM-L6-v2` model, so document embeddings do not require another API key.

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully.")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully.")
```

### ChromaDB

ChromaDB stores document embeddings and enables similarity-based retrieval.

The knowledge-base pipeline is:

PDF Documents
↓
PyPDFLoader
↓
Text Splitter
↓
Document Chunks
↓
Embeddings
↓
ChromaDB

```python
from langchain_chroma import Chroma

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory="chroma_db",
    collection_name="novatech_knowledge_base"
)

print("ChromaDB vector store created successfully.")
```

### Retriever

A retriever finds the most relevant document chunks for a user's question.

We retrieve the top 3 most similar chunks from ChromaDB.

```python
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)

print("Retriever created successfully.")
```

### Retrieval-Augmented Generation (RAG)

Retrieval-Augmented Generation (RAG) combines information retrieved from the company knowledge base with the language model.

Instead of asking the LLM to answer from its general knowledge, relevant company document chunks are retrieved first and provided as context to the model.

The RAG flow is:

User Question
↓
Retriever
↓
Relevant Document Chunks
↓
Prompt
↓
OpenRouter LLM
↓
Grounded Answer

### RAG Prompt

The model is instructed to answer only using information retrieved from the NovaTech Solutions knowledge base.

If the required information is not present in the retrieved documents, the model should clearly state that the information is not available rather than inventing a company policy.

```python
from langchain_core.prompts import ChatPromptTemplate

rag_prompt = ChatPromptTemplate.from_template("""
You are an internal AI assistant for NovaTech Solutions Pvt. Ltd.

Answer the employee's question using ONLY the provided company documents.

If the answer cannot be found in the provided documents, clearly say that the information is not available in the company knowledge base.

Do not invent company policies.

Context:
{context}

Employee Question:
{question}
""")

print("RAG prompt created successfully.")
```

```python
def format_docs(docs):
    return "\n\n".join(
        doc.page_content for doc in docs
    )

rag_chain = (
    {
        "context": retriever | format_docs,
        "question": lambda x: x
    }
    | rag_prompt
    | llm
)

print("RAG chain created successfully.")
```

### RAG Testing

The RAG system was tested using questions related to leave, work-from-home, working hours, and information that is not present in the knowledge base.

The system retrieves relevant company policy chunks and provides them as context to the OpenRouter LLM. For questions outside the available knowledge base, the system is instructed not to invent company policies.

```python
# Rag test code

def ask_rag(question):
    response = rag_chain.invoke(question)

    retrieved_docs = retriever.invoke(question)

    print("Question:")
    print(question)

    print("\nAnswer:")
    print(response.content)

    print("\nSources:")

    for doc in retrieved_docs:
        print("-", Path(doc.metadata["source"]).name)
```

### 10.1 HR Agent

The HR Agent is responsible for answering employee questions related to leave, attendance, work from home, and general employee policies.

It uses the existing company knowledge base and RAG system to provide answers grounded in NovaTech Solutions Pvt. Ltd. documents.

```python
from langchain_core.prompts import ChatPromptTemplate

hr_prompt = ChatPromptTemplate.from_template("""
You are the HR Assistant for NovaTech Solutions Pvt. Ltd.

Your responsibility is to answer employee questions related to:

- Leave
- Attendance
- Work From Home
- Employee policies
- Employee handbook information

Use the provided company knowledge base as your primary source.

IMPORTANT RULES:

1. Answer using the retrieved company documents.
2. Do not invent company policies.
3. If the information is not available in the retrieved documents, clearly state that the information is not available in the company knowledge base.
4. Keep answers clear, professional, and concise.
5. When appropriate, mention the source document used.

Retrieved Company Information:
{context}

Employee Question:
{question}
""")

print("HR prompt created successfully.")
```

```python
# HR Agent

def hr_agent(question):
    retrieved_docs = retriever.invoke(question)

    context = "\n\n".join(
        doc.page_content for doc in retrieved_docs
    )

    chain = hr_prompt | llm

    response = chain.invoke({
        "context": context,
        "question": question
    })

    sources = list(dict.fromkeys(
        Path(doc.metadata["source"]).name
        for doc in retrieved_docs
    ))

    return {
        "answer": response.content,
        "sources": sources
    }

print("HR agent function created successfully.")
```

```python
# Test HR Agent
def test_hr_agent(question):
    result = hr_agent(question)

    print("Question:")
    print(question)

    print("\nAnswer:")
    print(result["answer"])

    print("\nSources:")
    for source in result["sources"]:
        print("-", source)
```

### HR Agent test cases

```python
# tc 1
test_hr_agent(
    "What is the annual leave entitlement?"
)
# tc 2
test_hr_agent(
    "How many days in advance should I apply for planned leave?"
)
# tc 3
test_hr_agent(
    "How many days can an eligible employee work from home?"
)
# tc 4
test_hr_agent(
    "What are the standard working hours?"
)
# tc 5
test_hr_agent(
    "What are employees expected to do with confidential company information?"
)
# tc 6
test_hr_agent(
    "What is NovaTech's maternity leave policy?"
)
```

### HR Agent vs Basic RAG

In the previous RAG implementation, a question was directly passed to the retriever and then to the LLM.

The HR Agent adds a specialized role and instructions around the existing RAG system.

Basic RAG:

Question
↓
Retriever
↓
LLM
↓
Answer

HR Agent:

Employee HR Question
↓
HR Agent
↓
HR-specific instructions
↓
Retriever
↓
Company Documents
↓
LLM
↓
HR Answer

The HR Agent is therefore a specialized component designed specifically for employee and HR-related questions.
